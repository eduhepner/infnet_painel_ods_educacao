# Procurando e definindo os dados a serem buscados

```python
# Identificadores segundo https://servicodados.ibge.gov.br/api/docs/agregados?versao=3#api-acervo
# - 125: Taxa de escolarização
# - 126: taxa de alfabetização
# - 127: Escolaridade
# - 27: Educação
# - 302: Educação de qualidade
# - 235: Frequência escolar
# - 179: Instrução e educação

def consulta_assunto(assunto):
  url = f'https://servicodados.ibge.gov.br/api/v3/agregados?assunto={assunto}'
  response = requests.get(url)
  response.raise_for_status()
  dados = response.json()
  return dados

def print_assuntos(dados):
  for linha in dados:
    nome = linha['nome']
    for agregado in linha['agregados']:
      print(f'{assunto} - {nome}: {agregado}')
```

```python
def verificar_metadados(agregado):
  url = f'https://servicodados.ibge.gov.br/api/v3/agregados/{agregado}/metadados'
  response = requests.get(url)
  response.raise_for_status()
  dados = response.json()
  return dados

def print_metadados(dados):
  for linha in dados:
    if linha != 'classificacoes':
      print(f'{linha}: {dados[linha]}')
    else:
      for classificacao in dados[linha]:
        for categorias in classificacao['categorias']:
          print(f"classificacoes: {classificacao['id']} - {classificacao['nome']}: {categorias}")
```

**7138 — escolarização por sexo**

```python
agregado = 7138
metadados = verificar_metadados(agregado)
print_metadados(metadados)
```

```python
ano = metadados['periodicidade']['fim']
variavel = metadados['variaveis'][0]['id']

params = {
    "localidades": "N3[all]",
    "classificacao": "2[6794,4,5]|58[95253,41126,99749,47813,31615,2792,100052,108866]"
}

url = f"https://servicodados.ibge.gov.br/api/v3/agregados/{agregado}/periodos/{ano}/variaveis/{variavel}"
resp = requests.get(url, params=params, timeout=20)
resp.raise_for_status()
dados = resp.json()
```

```python
for linha in dados[0]['resultados']:
  print(linha)
```

```python
print(dados[0]["resultados"][0].keys())
print(dados[0]["resultados"][0]['classificacoes'][0].keys())
print(dados[0]["resultados"][0]['series'][0].keys())
```

```python
linhas = []

for resultado in dados[0]["resultados"]:
    # ------------------------------
    # Sexo
    # ------------------------------
    sexo = resultado["classificacoes"][0]["categoria"]

    # ------------------------------
    # Grupo de idade
    # ------------------------------
    grupo_idade = resultado["classificacoes"][1]["categoria"]

    # ------------------------------
    # Localidade
    # ------------------------------
    for serie in resultado["series"]:
        estado = serie["localidade"]["nome"]

        # ------------------------------
        # Valor do último ano
        # ------------------------------
        valor = serie["serie"][str(ano)]

        # ------------------------------
        # Montando as linhas
        # ------------------------------
        linhas.append({
            "Estado": estado,
            "Sexo": list(sexo.values())[0],
            "Grupo de idade": list(grupo_idade.values())[0],
            "Taxa de escolarização (%)": valor
        })
# ==========================================
# 4. DATAFRAME
# ==========================================
df = pd.DataFrame(linhas)

# ==========================================
# 5. CONVERSÃO DO VALOR
# ==========================================
df["Taxa de escolarização (%)"] = pd.to_numeric(
    df["Taxa de escolarização (%)"],
    errors="coerce"
)

# ==========================================
# 6. ORDENAÇÃO
# ==========================================
df = df.sort_values(
    by=["Estado", "Sexo"]
).reset_index(drop=True)

# ==========================================
# 7. VISUALIZAÇÃO
# ==========================================
print("\nQuantidade de linhas:", len(df))
print("\nEstados encontrados:", df["Estado"].nunique())
df
```
